Palabras reservadas:
    Son palabras y sintaxis reservadas para el uso de operaciones internas en python
    Ej: Def[para crear funciones], True[Para dar un valor verdadero], break[Para salir de un ciclo for, while]

Identificadores:
    Son los nombres que se le dan a Objetos en python, Clases, Funciones, Variables, Modulos. Se tienen unas reglas para nombrarlas.
    Alfanumerico y puede tener guión bajo, no se puede tener otros caracteres especiales, ni palabras reservadas, no se puede inciar con numero.
    Ej: variable123, variable_123, variableNombrada

Caracteres especiales:
    Son símbolos que tienen un significado funcional, estructural o de control dentro del código y las cadenas de texto, y no se interpretan literalmente.
    Ej: !,@,#,-,\,/,ñ

Constantes:
    Son identificadores o variables cuyo valor está diseñado para permanecer inalterado durante la ejecución de un programa.
    Ej: VALOR_PI, IVA

Variables:
    Son un nombre simbólico que actúa como referencia o etiqueta para un objeto (dato) almacenado en la memoria del ordenador.
    Ej: edadUsuario.

 Expresiones:
    son combinaciones de valores, variables, operadores y llamadas a funciones que el intérprete evalúa para producir un nuevo valor.
    Ej: 2 + 3, 1 > 2, and, or.

Instrucciones:
    son comandos o sentencias individuales que forman un programa y le ordenan a la computadora ejecutar acciones específicas.
    Ej: x = 10, print("Hello World").

Bucles:
    son estructuras de control que repiten un bloque de código varias veces, automatizando tareas repetitivas y mejorando la eficiencia del programa, componentes clave [break] [continue]
    Ej: For, While

Contadores:
    son herramientas utilizadas para llevar la cuenta de repeticiones de elementos.
    Ej: contador[a,b,c]

Acumuladores:
    es una variable utilizada dentro de un bucle (for o while) para almacenar resultados intermedios o totales, incrementando (o decrementando) su valor en cada iteración.
    Ej: i, j

Interruptores:
    son estructuras de control de flujo utilizadas para ejecutar diferentes bloques de código según el valor de una variable, simplificando múltiples condiciones.
    Ej:match-case, if-elif-else.

Estructuras:
    secuenciales:
        son la forma más básica de control de flujo, donde las instrucciones se ejecutan una tras otra en orden lineal, de arriba hacia abajo. 
        Ej:# 1. Entrada de datos
                lado = float(input("Ingrese el lado del cuadrado: "))

                # 2. Operación secuencial (cálculo)
                perimetro = lado * 4

                # 3. Salida de datos
                print("El perímetro es:", perimetro)

    repetitivas:
        son herramientas de control que permiten ejecutar un mismo bloque de código varias veces consecutivas sin tener que escribirlo de nuevo.
        Ej: for, while.

    selectivas:
        son instrucciones de control de flujo que permiten al programa tomar decisiones y ejecutar diferentes bloques de código basándose en la evaluación de condiciones lógicas (verdadero o falso).
        Ej: if, else, elif
